## 3. Dataset Inclusion Criteria

This study is restricted to runs satisfying the following criteria:
- Character chosen: Ironclad (card selection varies based on each character)
- Ascension level (10-20):  Higher ascension levels encourage more optimized decision-making, making them more representative of experienced play.
- No daily runs: daily runs have special rules that affect the card choices
- No special seeds: same reasoning with daily runs
- No chosen seeds: players may have prior knowledge before deciding if they know the seed
- No endless runs: they have special rules that affect the card choices in a non meaningful way

### Assumptions

A1. One row represents one card-selection decision.

A2. The target variable is the chosen card.

A3. Offered card ordering has no semantic meaning.

A4. The raw dataset preserves the original ordering only because it mirrors the source logs.

A5. Feature engineering will remove this ordering before modeling.

A6. The modeling dataset (long format) is derived from the canonical dataset via a documented transformation and is never edited independently.

## 7. Snapshot Principle

Every observation represents the complete game state immediately before the player makes a card-selection decision. Therefore, every feature in the dataset must correspond to information that was available to the player at that exact moment. Any information generated after the decision (for example, the chosen card being added to the deck or the eventual outcome of the run) must not appear in the observation.
